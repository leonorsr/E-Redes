```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Energia Produzida Total Nacional

```python
# Carregar o arquivo CSV com o separador correto
dados = pd.read_csv('./data/energia-produzida-total-nacional.csv', sep=';')

# Filtrar os dados para os anos de 2023 e 2024
dados['Ano'] = pd.to_datetime(dados['Data']).dt.year
dados['Mês'] = pd.to_datetime(dados['Data']).dt.month

dados_filtrados = dados[dados['Ano'].isin([2023, 2024])]

# Agrupar por 'Ano' e 'Mês', somando as colunas 'Total (kWh)', 'Mercado (kWh)', 'Regime Especial (kWh)'
dados_agrupados = dados_filtrados.groupby(['Ano', 'Mês'])[['Total (kWh)', 'Mercado (kWh)', 'Regime Especial (kWh)']].sum().reset_index()

# Criar uma coluna de "Data" para facilitar a plotagem contínua
dados_agrupados['Data'] = pd.to_datetime(dados_agrupados['Ano'].astype(str) + '-' + dados_agrupados['Mês'].astype(str) + '-01')

# Criar a figura com 3 subgráficos lado a lado
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Gráfico para 'Total (kWh)'
axes[0].plot(dados_agrupados['Data'], dados_agrupados['Total (kWh)'], marker='o', color='b', label='Energia Total (kWh)')
axes[0].set_title('Evolução Mensal da Energia Total (kWh) - 2023 e 2024', fontsize=14)
axes[0].set_xlabel('Data', fontsize=12)
axes[0].set_ylabel('Energia Total (kWh)', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True)
axes[0].legend()

# Gráfico para 'Mercado (kWh)'
axes[1].plot(dados_agrupados['Data'], dados_agrupados['Mercado (kWh)'], marker='o', color='g', label='Mercado (kWh)')
axes[1].set_title('Evolução Mensal da Energia Mercado (kWh) - 2023 e 2024', fontsize=14)
axes[1].set_xlabel('Data', fontsize=12)
axes[1].set_ylabel('Mercado (kWh)', fontsize=12)
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True)
axes[1].legend()

# Gráfico para 'Regime Especial (kWh)'
axes[2].plot(dados_agrupados['Data'], dados_agrupados['Regime Especial (kWh)'], marker='o', color='r', label='Regime Especial (kWh)')
axes[2].set_title('Evolução Mensal da Energia Regime Especial (kWh) - 2023 e 2024', fontsize=14)
axes[2].set_xlabel('Data', fontsize=12)
axes[2].set_ylabel('Regime Especial (kWh)', fontsize=12)
axes[2].tick_params(axis='x', rotation=45)
axes[2].grid(True)
axes[2].legend()

# Ajustar o layout para que os gráficos não se sobreponham
plt.tight_layout()

# Exibir o gráfico
plt.show()
```

## Evolução Mensal (2023-2024)

```python
# Carregar o arquivo CSV com o separador correto
dados = pd.read_csv('./data/energia-produzida-total-nacional.csv', sep=';')

# Filtrar os dados para os anos de 2023 e 2024
dados['Ano'] = pd.to_datetime(dados['Data']).dt.year
dados['Mês'] = pd.to_datetime(dados['Data']).dt.month

dados_filtrados = dados[dados['Ano'].isin([2023, 2024])]

# Agrupar por 'Ano' e 'Mês', somando a coluna 'Total (kWh)'
dados_agrupados = dados_filtrados.groupby(['Ano', 'Mês'])['Total (kWh)'].sum().reset_index()

# Criar uma coluna de "Data" para facilitar a plotagem contínua
dados_agrupados['Data'] = pd.to_datetime(dados_agrupados['Ano'].astype(str) + '-' + dados_agrupados['Mês'].astype(str) + '-01')

# Plotar o gráfico de evolução temporal
plt.figure(figsize=(10, 6))
plt.plot(dados_agrupados['Data'], dados_agrupados['Total (kWh)'], marker='o', color='b', label='Energia Total (kWh)')
plt.title('Evolução Mensal da Energia Total (kWh) - 2023 e 2024', fontsize=14)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Energia Total (kWh)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
