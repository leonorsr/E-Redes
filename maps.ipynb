```python
import geopandas as gpd

# Carregar o arquivo GeoJSON
mapa_portugal = gpd.read_file("pt.json")

# Exibir as primeiras linhas do GeoDataFrame para verificar se foi carregado corretamente
print(mapa_portugal.head())
```

```python
import matplotlib.pyplot as plt

# Visualizar o mapa de Portugal
mapa_portugal.plot(figsize=(10, 10))

# Ajuste do título
plt.title("Mapa de Portugal - Divisão Administrativa", fontsize=16)

# Exibir o mapa
plt.show()
```
