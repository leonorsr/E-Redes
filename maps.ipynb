```python
import geopandas as gpd
import matplotlib.pyplot as plt

# Carregar o shapefile (substitua pelo caminho correto do arquivo .shp)
shapefile_path = "distritos.shp"  # Certifique-se de ter o arquivo  correto
mapa = gpd.read_file(shapefile_path)

# Plotar o mapa
fig, ax = plt.subplots(figsize=(10, 10))
mapa.plot(ax=ax, edgecolor='black', cmap='Blues', alpha=0.7)

# Adicionar nomes dos distritos
for x, y, label in zip(mapa.geometry.centroid.x, mapa.geometry.centroid.y, mapa["NOME"]):
    ax.text(x, y, label, fontsize=8, ha='center', color='black')

plt.title("Mapa de Portugal - Divisão por Distritos")
plt.axis("off")
plt.show()
```
