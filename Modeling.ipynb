## **Modelos Clássicos de Séries Temporais**

### **ARIMA / SARIMA**  
- Ideal para séries **com padrão e sem tendência clara**
- **SARIMA** se tiver **padrões sazonais** (tipo picos diários ou semanais).
- São **fáceis de interpretar** e rápidos de treinar.

> Se o consumo tem padrões repetitivos (tipo todos os dias às 19h há pico), **SARIMA** pode capturar isso muito bem.

- Funciona para séries **com tendência, sazonalidade, feriados** etc.
