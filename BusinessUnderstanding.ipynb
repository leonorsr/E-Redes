# **Business Understanding**

## **Objetivos**




O objetivo do projeto é prever o consumo de energia elétrica da rede. Esta previsão é fundamental para uma série de atividades dentro das empresas de energia, tais como:

* Antecipar a demanda de energia evitando sobrecargas ou faltas de energia na rede.

* Garantir que a geração e a distribuição de energia sejam feitas de forma eficiente.

* Previsão de lucros na distribuição de energia, ajustando os preços e políticas de cobrança.

* Identificação padrões sazonais ou anormais no consumo e ajuste nas estratégias de operação.

## **Coleta e Compreensão dos Dados**

Características dos dados disponíveis no *dataset* *consumo-total-nacional.csv*:

* Data/Hora: A marca temporal do consumo (formato de data e hora).

* Baixa Tensão (kWh): Consumo de energia na rede de baixa tensão.

* Média Tensão (kWh): Consumo de energia na rede de média tensão.

* Alta Tensão (kWh): Consumo de energia na rede de alta tensão.

* Muito Alta Tensão (kWh): Consumo de energia na rede de muito alta tensão.

* Total (kWh): Consumo total de energia.

Perguntas para entender os dados:

* Existe uma variação significativa no consumo ao longo das semanas/meses/anos?

* O consumo é maior em horários específicos ou dias da semana?

* Quais são os picos de demanda, e quando ocorrem?

* Como o consumo de energia é distribuído entre as diferentes faixas de tensão (baixa, média, alta, muito alta)?
