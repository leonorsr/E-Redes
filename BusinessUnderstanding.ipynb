# Business Understanding

## Objetivos




### 1. **Modelos Clássicos de Séries Temporais (Time Series Stamps)**

Estes são os modelos mais usados para séries estacionárias:

#### 🔹 **ARIMA / SARIMA**  
- Ideal para séries **com padrão e sem tendência clara**, como a tua.
- **SARIMA** se tiver **padrões sazonais** (tipo picos diários ou semanais).
- São **fáceis de interpretar** e rápidos de treinar.
- Ótimos para começar: se funcionarem bem, talvez nem precises complicar.

> 🔧 Se o teu consumo tem padrões repetitivos (tipo todos os dias às 19h há pico), **SARIMA** pode capturar isso muito bem.

#### 🔹 **Prophet (do Facebook)**  
- Funciona para séries **com tendência, sazonalidade, feriados** etc.
- Mais indicado se os dados não fossem estacionários, mas podes experimentar — é fácil de usar e configurar.
- Menos flexível que ARIMA/SARIMA para séries puramente estacionárias.

---

### 2. **Redes Neuronais (Deep Learning)**

Usar **redes neuronais** para séries temporais pode ser interessante **se quiseres ir além**, especialmente se:

- Quiseres prever com **várias variáveis externas** (ex: temperatura, hora do dia, dia da semana...).
- A série tiver **padrões não lineares** que modelos clássicos não capturam.
- Tiveres muitos dados.

#### 🔸 **MLP (Multilayer Perceptron)**
- Simples, mas pode funcionar bem para **padrões curtos e limpos**.
  
#### 🔸 **LSTM / GRU (redes recorrentes)**
- Ideais para capturar **dependências de longo prazo** (ex: o consumo de hoje é afetado por padrões de dias anteriores).
- Mais robustas e poderosas, mas também **mais difíceis de treinar**.
  
#### 🔸 **CNNs para séries temporais**
- Sim, CNN também funcionam — capturam **padrões locais** (tipo sazonalidade diária).

> ⚠️ Mas lembra-te: redes neurais precisam de mais dados, mais tuning, e não são tão fáceis de explicar.

---

### 3. **Híbridos: o melhor dos dois mundos**

Por que não usar os dois?

- Usa **ARIMA/SARIMA** como baseline (modelo base).
- Depois, experimenta uma **rede neural** para ver se melhora a performance.
- Há até abordagens híbridas (ex: usar ARIMA para modelar a parte linear e uma LSTM para os resíduos, que são não-lineares).
